```python
##### How to Approach Machine Learning Projects #####

# In this tutorial, we'll explore a step-by-step process for approaching ML problems:
#    - Understand the business requirements and the nature of the available data.
#    - Classify the problem as supervised/unsupervised and regression/classification.
#    - Download, clean & explore the data and create new features that may improve models.
#    - Create training/test/validation sets and prepare the data for training ML models.
#    - Create a quick & easy baseline model to evaluate and benchmark future models.
#    - Pick a modeling strategy, train a model, and tune hyperparameters to achieve optimal fit.
#    - Experiment and combine results from multiple strategies to get a better result.
#    - Interpret models, study individual predictions, and present your findings.

#!pip install opendatasets --upgrade --quiet

#import os
#import jovian
#import matplotlib
#import opendatasets as od
#import numpy as np
#import pandas as pd
#import matplotlib.pyplot as plt
#import seaborn as sns
#import plotly.express as px
#%matplotlib inline

#sns.set_style('darkgrid')
#matplotlib.rcParams['font.size'] = 14
#matplotlib.rcParams['figure.figsize'] = (10, 6)
#matplotlib.rcParams['figure.facecolor'] = '#00000000'


### Step 1 - Understand Business Requirements & Nature of Data
# Most machine learning models are trained to serve a real-world use case. It's important to understand the business requirements, modeling objectives and the
# nature of the data available before you start building a machine learning model.

## Understanding the Big Picture
# The first step in any machine learning problem is to read the given documentation, talk to various stakeholders and identify the following:
#   - What is the business problem you're trying to solve using machine learning?
#   - Why are we interested in solving this problem? What impact will it have on the business?
#   - How is this problem solved currently, without any machine learning tools?
#   - Who will use the results of this model, and how does it fit into other business processes?
#   - How much historical data do we have, and how was it collected?
#   - What features does the historical data contain? Does it contain the historical values for what we're trying to predict.
#   - What are some known issues with the data (data entry errors, missing data, differences in units etc.)
#   - Can we look at some sample rows from the dataset? How representative are they of the entire dataset.
#   - Where is the data stored and how will you get access to it?
#   - ...

# Gather as much information about the problem as possible, so that you're clear understanding of the objective and feasibility of the project.
# This is a huge shift in perspective.

# Until now, you've been given a perfectly clean problem: "Here is a CSV file, here is the target column, go train a model."
# In the real world, nobody hands you a clean CSV. You have to walk into a room with a client, a manager, or a product team, and ask them these
# exact questions before you write a single line of code.
# If you skip this step, you might spend 3 weeks building a perfect ML model that solves the wrong problem.

# Let's break down the most important questions in this list and what they actually mean in the real world:

#   - "What is the business problem you're trying to solve using machine learning?"
# Why it matters: You need to define the exact objective and the target variable.
# Example: "We want to predict which customers will cancel their Netflix subscription next month based on their viewing history."

#   - "Why are we interested in solving this problem? What impact will it have on the business?"
# Why it matters: You need to understand the financial or operational value. Why does this matter?
# Example: "If we can predict who leaves, we can offer them a discount and save ₦10 million a year."

#   - "How is this problem solved currently, without any machine learning tools?"
# Why it matters: This establishes your baseline. If the current manual solution works 90% of the time, your model needs to beat 90% to be worth building.
# Example: If a hospital currently uses a simple rule ("If patient is over 65, they are high risk"), and it works 90% of the time, your model needs
# to beat that baseline to be worth the cost.

#   - "Who will use the results of this model, and how does it fit into other business processes?"
# Why it matters: It changes how you build the model.
# Example: If a doctor uses it, they need to see the feature importances (explainability). If an automated system uses it
# (like a credit card fraud detector), it just needs to be fast and accurate.

#   - "How much historical data do we have, and how was it collected?"
# Why it matters: You need to know if you have enough past examples for the model to learn from, and if the data is reliable.
# Example: "We have 3 years of customer data, collected from our website signup forms."

#   - "What features does the historical data contain? Does it contain the historical values for what we're trying to predict?"
# Why it matters: You are checking for the target variable and looking out for Data Leakage. You can only use data that would be available
# at the time of prediction.
# Example: Imagine predicting loan defaults. If your dataset has a column collection_agency_called, the model will use this to predict default perfectly.
# But you wouldn't know that before they defaulted! That's data leakage.

#   - "What are some known issues with the data (data entry errors, missing data, differences in units etc.)?"
# Why it matters: Real-world data is messy. You need to know about errors, missing values, or inconsistent units before you start coding.
# Example: "Oh yeah, the temperature sensor in Sydney was broken in 2015, so all those values are -999."

#   - "Can we look at some sample rows from the dataset? How representative are they of the entire dataset?"
# Why it matters: You need to visually inspect the data to ensure it isn't heavily biased or skewed toward one group.
# Example: "We looked at the first 10 rows, but they are all from New York. We need to make sure the data includes people from other states too."

#   - "Where is the data stored and how will you get access to it?"
# Why it matters: You need to figure out the logistics of getting the data into your Jupyter notebook.
# Example: "It's in an SQL database, and IT needs to give me read access."

# Why this matters for your portfolio:
# When you do your next Kaggle project (like the Titanic dataset), don't just open the CSV and run model.fit().

# Write a short "Big Picture" section at the top of your notebook:
# -> Business Problem: Predict who survived the Titanic to understand what factors saved lives.
# -> Current Solution: None. (It's a historical dataset).
# -> Target Audience: Historians or safety regulators.
# -> Known Issues: The Age column has 177 missing values. The Cabin column is mostly empty.
# That single markdown cell will make you look 10x more professional than 90% of people on GitHub.


# Note: The "Big Picture" step is about understanding the business context before coding. Key questions include:
# What is the problem? How is this solved currently? Who uses the model? Where does the data come from? And what are the known data issues?
# Skipping this leads to building models that are useless in the real world.


### Step 2 - Classify the problem as supervised/unsupervised & regression/classification
# This is a crucial step. You are now learning how to choose the right tool for the job before you even write a line of code.

##The Problem Classification Tree:
# Before you touch Python, you look at your problem and ask two questions:

# Question A: Do I have the answers in my data? (Supervised vs. Unsupervised)
# -> Supervised (Data is pre-categorized): You have the labels. You have the answer key.
#   (e.g., "Here are 1,000 medical records with a column saying if they got sick or not.")
# -> Unsupervised (Data is not labeled): You have no answer key. You just want to find patterns.
#   (e.g., "Here is a list of 1,000 customers. Group them into similar groups.")

# Question B: What is the output? (Regression vs. Classification)
# (If you are doing Supervised Learning)
# -> Classification: Predicting a Category (Yes/No, Spam/Not Spam, Dog/Cat).
# -> Regression: Predicting a Number (Price, Temperature, Age).

# Note: You have already done Supervised Classification (Rain Tomorrow), Supervised Regression (Medical Charges),
# and seen the theory behind Ensembling (Random Forests).


## Loss Functions and Evaluation Metrics
# Once you have identified the type of problem you're solving, you need to pick an appropriate evaluation metric.
# Also, depending on the kind of model you train, your model will also use a loss/cost function to optimize during the training process.
#  -> Evaluation metrics - they're used by humans to evaluate the ML model
#  -> Loss functions - they're used by computers to optimize the ML model
# They are often the same (e.g. RMSE for regression problems), but they can be different (e.g. Cross entropy and Accuracy for classification problems).

## You have not touched Unsupervised Learning yet.
# Everything you've done so far—Linear Regression, Logistic Regression, Decision Trees, and Random Forests—is Supervised Learning.
# You always gave the model the answers (the train_targets) so it could learn.
# Unsupervised Learning is a completely different ballgame. There is no answer key. You just give the model data and say, "Find patterns."

# Unsupervised Learning has three main pillars:
#   -> Clustering: Grouping similar things together.
# Example: You have 10,000 customers and no labels. The model groups them into "High spenders," "Bargain hunters," and "Window shoppers"
# based on their behavior.

#   -> Dimension Reduction: Squashing many columns into fewer columns.
# Example: You have 100 columns of customer data. The model squashes it down to 2 columns so you can plot it on a graph and see the patterns visually.

#   -> Association: Finding things that go together.
# Example: "Customers who buy bread also buy butter" (the classic Amazon recommendation engine logic).


## Deep Learning: Deep Learning (Neural Networks) is just a powerful tool that can do any of these (supervised or unsupervised).
#  But you don't need to worry about Deep Learning right now. You are mastering Classical ML first.


### Step 3 - Download, clean & explore the data and create new features
## Downloading Data
# There may be different sources to get the data:
#   - CSV files
#   - SQL databases
#   - Raw File URLs
#   - Kaggle datasets
#   - Google Drive
#   - Dropbox
#   - etc.
# Identify the right tool/library to get the data.

## Cleaning Data
# The first step is to check the column data types and identify if there are any null values.

## Exploratory Data Analysis and Visualization
# Objectives of exploratory data analysis:
# - Study the distributions of individual columns (uniform, normal, exponential)
# - Detect anomalies or errors in the data (e.g. missing/incorrect values)
# - Study the relationship of target column with other columns (linear, non-linear etc.)
# - Gather insights about the problem and the dataset
# - Come up with ideas for preprocessing and feature engineering

## Feature Engineering
# Feature engineer is the process of creating new features (columns) by transforming/combining existing features or by incorporating data
# from external sources.
# For example, here are some features that can be extracted from the "Date" column:
# - Day of week
# - Day or month
# - Month
# - Year
# - Weekend/Weekday
# - Month/Quarter End

# What they did: They took the Date column (which is just text like "2013-01-01") and split it into Day, Month, and Year.
# Why it matters: A model can't understand "2013-01-01". But it can understand "Month = 1" (January) and "Day = 1".
# This lets the model learn things like "Sales are always high in December" or "Sales drop in January."


### Step 4 - Create a training/test/validation split and prepare the data for training
# The process of preparing data for a model follows the exact same workflow regardless of the dataset:
#  1. Train/Validation/Test Split: For time-series data, always split by date (train on the past, validate/test on the future) to prevent data leakage.
#  2. Identify Columns: Separate your inputs (X) from your target (y).
#     Crucial Rule: Never include columns that wouldn't be available at the time of prediction (e.g., "Number of Customers" or "Collection Agency Called").

#     Here is a deeper explanation of that "Crucial Rule" (Data Leakage).
#     Think of it like trying to predict the winner of a football match before the game starts.
#
#     # The Problem: Time Traveling Data
#     If you use data from after the game starts to predict the winner, your model is cheating. It's using time-traveling information.
#     In Machine Learning, this is called Data Leakage, and it is the #1 reason models fail in the real world.
#
#     -> "Number of Customers"
#      The Goal: You want to predict today's sales before the store opens.
#      The Trap: The dataset has a column for Customers (how many people walked in). It has a 0.89 correlation with Sales! If you give this to the model,
#      it will say: "Ah, 500 customers walked in, so sales will be ₦50,000."
#      Why it fails: When you deploy this model to a real store manager at 8:00 AM, you don't know how many customers will walk in today.
#      You can't predict the future. The model is useless.
#
#     -> "Collection Agency Called"
#      The Goal: You want to predict if a customer will default on a loan.
#      The Trap: The dataset has a column for Collection Agency Called. If the agency called, the customer definitely defaulted.
#      Why it fails: By the time the collection agency calls, the loan has already defaulted. You are trying to predict an event that has already happened.
#      You need data from before the loan was approved to make a useful prediction.
#
#      How to avoid it:
#      Ask yourself: "At the exact moment I need to make this prediction, will I actually have this piece of information?"
#      - Predicting Rain Tomorrow: Do I know today's humidity? Yes. (Safe to use).
#      - Predicting Medical Bills: Do I know the patient's age and BMI? Yes. (Safe to use). Do I know the final hospital bill? No. (Leakage).
#      - Predicting Store Sales: Do I know the date and if a promo is running? Yes. (Safe to use). Do I know the number of customers? No. (Leakage).
#
#      Note: Data leakage happens when you train a model using data that won't be available when the model is actually used
#      (e.g., predicting today's sales using today's customer count). Always ask: "Will I have this information at the exact moment
#      I need to make a prediction?" If the answer is no, drop the column.
#
#      If you can't use those "leaky" columns, how do you actually solve the problem?
#      You have to rely on predictive features (things you do know in advance) instead of outcome features (things that happen later).
#      Here is how you counter data leakage in those exact scenarios:
#        -> The Store Sales Case (Predicting Sales)
#        The Leak: You can't use Customers (how many people actually walked in).
#        The Counter: You use historical averages and external data.
#         "In the past, when there was a Promo on a Tuesday in December, we averaged 5,000 customers."
#         "The weather forecast says it's raining tomorrow, so foot traffic will likely be lower."
#        You use features you know before the store opens (Day of week, Promo, Holiday, Weather forecast) to estimate the customers,
#        and then estimate the sales.
#
#       -> The Loan Default Case (Predicting Defaults)
#       The Leak: You can't use Collection Agency Called (happens after default).
#       The Counter: You use applicant data at the time of application.
#         Credit score, income, debt-to-income ratio, employment history, age, and previous repayment history on other loans.
#         You build a model that predicts the risk before you give them the money.
#
#       -> The Medical Bills Case (Predicting Cost)
#       The Leak: You can't use Final Bill Amount or Length of Stay (you only know these after the patient is treated).
#       The Counter: You use intake data.
#         Age, BMI, smoking status, pre-existing conditions, and family history. You predict the risk of a high bill before the patient gets admitted.
#
#        The Technical Countermeasures (How Data Scientists prevent it)
#       If you are working with a massive dataset and you're afraid of accidental leakage, here are the 3 technical rules you follow:
#       -> The "Time-Machine" Test:
#        Always split your data by time (like you did with the Rain dataset). If you train on 2013-2014 and test on 2015, you prevent the model from
#        accidentally learning patterns from the future.
#       -> Drop the ID and Target Correlations:
#        If a column is basically a proxy for the target (like Collection Agency Called for Loan Default), delete it immediately.
#        You can find these by looking at the correlation matrix—if a feature has a 0.99 correlation with the target, it's probably leakage.
#       -> Build the Preprocessing Pipeline First:
#        When you use Scikit-learn's Pipeline and ColumnTransformer, you fit the scaler/imputer only on the training data, then transform the validation/test data. This prevents "data peeking."
#
#       # Note: Correlation is a great tool, but a correlation of 0.99 is a red flag for Data Leakage. If a feature perfectly predicts the target,
#               it's usually because it happens after the event (like a collection agency call after a default) or is the exact same thing as the target
#               (like money in the register vs. sales).
#
#
#  3. Numeric vs. Categorical:
#     -> Use numeric columns for things with natural order (Age, Month, Year).
#     -> Use categorical/one-hot encoding for things without order (Day of the Week, Store Type, Location).
#
#  4. Preprocessing Pipeline:
#     -> Impute: Fill missing numeric values with the mean or median.
#     -> Scale: Use MinMaxScaler to squash numbers into a 0-1 range.
#     -> Encode: Use OneHotEncoder to turn text categories into 0s and 1s
#
# 5. Final Extraction: Combine the numeric and encoded columns into your final X_train, X_val, and X_test variables.
#



### Step 5 - Create quick & easy baseline models to benchmark future models
# A quick baseline model helps establish the minimum score any ML model you train should achieve.
# This is a masterclass in how real Data Scientists work.
# You just learned that you should never start with a complex model like a Random Forest or XGBoost.
# You start with the dumbest possible models to set a baseline. If your fancy AI can't beat a dumb guess, your AI is useless.
# Let's break down the 3 baselines they tested::

##  1. The "Dumb" Baseline (Always predict the mean)
#def return_mean(inputs):
#    return np.full(len(inputs), merged_df.Sales.mean())
#train_preds = return_mean(X_train)
#train_preds
# Let's evaluate this to using the RMSE score.
#from sklearn.metrics import mean_squared_error
#mean_squared_error(train_preds, train_targets, squared=False)
# output: 3082.450443277419
#mean_squared_error(return_mean(X_val), val_targets, squared=False)
# output: 3168.6033635047716

# What it does: It ignores all the input data (store, promo, day of week) and just predicts the average sales for every single day.
# The Result: RMSE = **$3082** (It's off by about $3000 on average).

## 2. The "Random" Baseline (Guess a random number)
# Let's try another model, which makes a random guess between the lowest and highest sale.
#def guess_random(inputs):
#    lo, hi = merged_df.Sales.min(), merged_df.Sales.max()
#    return np.random.random(len(inputs)) * (hi - lo) + lo
#train_preds = guess_random(X_train)
#train_preds
#mean_squared_error(guess_random(X_val), val_targets, squared=False)
# ouput: 18383.589238754674

# What it does: It guesses a random number between the minimum and maximum sales.
# The Result: RMSE = **$18315** (It's off by $18,000! It's terrible).

## 3. The "Simple ML" Baseline (Linear Regression)
# Let's train a simple LinearRegression model, with no customization.
#from sklearn.linear_model import LinearRegression
#linreg = LinearRegression()
#linreg.fit(X_train, train_targets)
#train_preds = linreg.predict(X_train)
#train_preds
#mean_squared_error(train_preds, train_targets, squared=False)
# output: 2741.5918405767957

#val_preds = linreg.predict(X_val)
#val_preds
#mean_squared_error(val_preds, val_targets, squared=False)
# output: 2817.5187414380916

# What it does: A basic straight line using all the features. No tuning, no hyperparameters.
# The Result: RMSE = **$2817** (Off by about $2,800).

# The Benchmark
# The lesson says: "Based on the above baselines, we now know that any model we train should ideally have an RMSE score lower than $2800."
# This is your target. If you train a Random Forest and it gets an RMSE of $3500, you throw it in the trash. It's worse than a simple straight line!


### Step 6 - Pick a strategy, train a model & tune hyperparameters
# Systematically Exploring Modeling Strategies
# Here's the general strategy to follow:
#  1. Find out which models are applicable to the problem you're solving.
#  2. Train a basic version for each type of model that's applicable
#  3. Identify the modeling approaches that work well and tune their hypeparameters
#  4. Use a spreadsheet to keep track of your experiments and results.

# Let's define a function try_model, which takes a model, then performs training and evaluation.
#def try_model(model):
    # Fit the model
#    model.fit(X_train, train_targets)

    # Generate predictions
#    train_preds = model.predict(X_train)
#    val_preds = model.predict(X_val)

    # Compute RMSE
#    train_rmse = mean_squared_error(train_targets, train_preds, squared=False)
#    val_rmse = mean_squared_error(val_targets, val_preds, squared=False)
#    return train_rmse, val_rmse

## Linear Models
#from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet, SGDRegressor

#try_model(LinearRegression())
# output: (2741.5918405767957, 2817.5187414380916)

# try_model(Ridge())
# ouput: (2741.5871950814444, 2817.7784665409868)

#try_model(Lasso())
# output: (2741.7143904640566, 2817.9472924045217)

#try_model(ElasticNet())
# output: (2879.0432103903954, 2972.6658860503385)

#try_model(SGDRegressor())
# output: (2741.790249222216, 2819.390224266075)

## Note: Standard Linear Regression overfits when there are many features. We use variations to fix this:
# -> Ridge: Shrinks weights (L2) to prevent overfitting.
# -> Lasso: Shrinks useless weights to exactly zero (L1) to select features.
# -> ElasticNet: Combines Ridge and Lasso.
# -> SGDRegressor: Uses gradient descent, which is faster for massive datasets.

# Look at the RMSE scores, they are all basically the same. Why? Because the data is highly complex (store sales change wildly based on holidays,
# weekends, and seasons), and a straight line can't capture those patterns.


## Tree Based Models
#from sklearn.tree import DecisionTreeRegressor, plot_tree
#tree = DecisionTreeRegressor(random_state=42)
#try_model(tree)
# output: (0.0, 1559.7378600480247)

# Why 0.0? Because an un-tuned Decision Tree memorized the training data perfectly (overfitting!).
# Why 1559? Even though it overfitted, it still cut the validation error almost in half compared to Linear Regression.

# Let's try a random forest.
#from sklearn.ensemble import RandomForestRegressor
#rf = RandomForestRegressor(random_state=42, n_jobs=-1)
#try_model(rf)
# output: (474.9198974666898, 1371.7321923469985)

# Why is Training 474? Because a Random Forest is made of 100 trees that average their predictions, it doesn't get a perfect 0 on training (which is good).
# Why is Validation 1371? It's the best score yet! It beats the decision tree and destroys the linear models.

# The Big Lesson: "The Spreadsheet"
# The lesson says: "Use a spreadsheet to keep track of your experiments and results."
# If you were doing this for a job, your spreadsheet would look like this:

# Model	               Train RMSE           	Val RMSE            	Notes
# Baseline (Mean)	        3082	                3168	             Useless
# Linear Regression	      2741	                2817	         Okay, but not great
# Decision Tree	          0.0                 	1559	         Overfit, but better
# Random Forest	          474                  	1371	           Best so far!

# Your new benchmark: You now know that any future model must get an RMSE lower than 1371 to be worth using.

## Note: When testing modeling strategies, use a spreadsheet to track the RMSE of each model. Linear models all performed similarly (~2817).
#  The Decision Tree overfit (Train RMSE 0.0, Val 1559). The Random Forest cut the error in half (Val RMSE 1371) and is the new best baseline.


### Step 7 - Experiment and combine results from multiple strategies
# In general, the following strategies can be used to improve the performance of a model:
#   -> Gather more data. A greater amount of data can let you learn more relationships and generalize the model better.
#   -> Include more features. The more relevant the features for predicting the target, the better the model gets.
#   -> Tune the hyperparameters of the model. Increase the capacity of the model while ensuring that it doesn't overfit.
#   -> Look at the specific examples where the model make incorrect or bad predictions and gather some insights
#   -> Try strategies like grid search for hyperparameter optimization and K-fold cross validation
#   -> Combine results from different types of models (ensembling), or train another model using their results.

# This is the God-Tier level of Machine Learning

#   1. Grid Search (Automated Tuning)
#      Remember when you manually tested max_depth=5 and max_depth=7 in the Random Forest? That was manual tuning.
#      Grid Search does it automatically. You give it a list of options:
#      grid = {'max_depth': [10, 20, 30], 'n_estimators': [100, 500]}
#      The computer will train 6 different models (3x2) and tell you exactly which combination won. It saves you hours of guessing.

#   2. K-Fold Cross Validation (The Ultimate Test)
#      Remember when we split the data into Train and Validation? What if we got lucky with the split? What if the validation set just happened
#      to be really easy to predict?
#      K-Fold fixes this. It takes the training data and chops it into 5 equal pieces (Folds).
#        . Round 1: Trains on folds 1,2,3,4. Tests on fold 5.
#        . Round 2: Trains on 1,2,3,5. Tests on fold 4.
#        . It does this 5 times and averages the scores.
#      If your model gets 85% across all 5 different tests, you know it's a solid model, not just a lucky one.

#   3. Ensembling and Stacking
#      Ensembling refers to combining the results of multiple models.
#      Stacking is a more advanced version of ensembling, where we train another model using the results from multiple models.
#
#      You already did Ensembling with the Random Forest (averaging 100 Decision Trees).
#      Stacking is the next level. Instead of just averaging, you train a brand new model (a Meta-Model) to figure out the best way to combine the predictions.
#      Imagine you have a Linear Regression model, a Random Forest, and an XGBoost model.
#      You ask them to predict store sales.
#      Then, you take their 3 predictions and feed them into a final "boss" model (like a simple Linear Regression).
#      The boss model learns: "The Random Forest is usually right, but when it disagrees with XGBoost, I should trust XGBoost."

## The Catch (The Cost of Power)
# All of these strategies require massive computational power.
# A Grid Search with K-Fold on a 1-million-row dataset can take hours or even days to run on a normal laptop.
# This is why companies use cloud servers with 32 CPU cores.

## Note:
#  -> Grid Search: Automates hyperparameter tuning by testing every combination.
#  -> K-Fold Cross Validation: Splits data into 5 chunks, trains 5 times, and averages the score to ensure the model isn't just lucky.
#  -> Stacking: Trains a final "meta-model" to learn the best way to combine predictions from multiple different models.



### Step 8 - Interpret models, study individual predictions & present your findings
# Feature Importance
# You'll need to explain why your model returns a particular result. Most scikit-learn models offer some kind of "feature importance" score.

# You have made it to the absolute final step of the ML lifecycle: Step 8. This is where you take your model out of the lab and show it to the world.
# Let's break down what this block is doing, because it's what separates a "coder" from a "Data Scientist."

# Once a model is trained and tested, the final step is to explain its results and present them to stakeholders.
#rf.feature_importances_

#  1. Feature Importance (The "Why")
# - Most Scikit-learn models assign an "importance" score to each feature.
# - Why it matters: You use this to explain the model's logic to non-technical people. Instead of saying "The model has an RMSE of 1371,"
#   you say "The model relies most heavily on whether a promotion is running, followed by the day of the week."
#   The Explainability Trade-off: Random Forests are "black boxes" (hard to explain exactly why a specific prediction was made). If a stakeholder needs 100%
#   transparency, you can use a single Decision Tree instead, which shows a clear flowchart of Yes/No decisions (at the cost of slightly lower accuracy).

#  2. Studying Individual Predictions
# - Look at specific rows where the model was wrong.
# - Business Rules: Sometimes you combine ML with manual logic. For example, if a store is closed, you hard-code the prediction to 0 instead
#   of asking the ML model.

#  3. Presenting to Stakeholders
# - No Code: Never show code or technical jargon to a business audience.
# - Focus on Value: Use visualizations (like feature importance charts). Focus on metrics that matter to the business
#   (e.g., "This model reduces inventory waste by 15%").
# - Be Honest: Explain the model's limitations and how it can be improved over time.

# More:
#  - Create a presentation for non-technical stakeholders
#  - Understand your audience - figure out what they care about most
#  - Avoid showing any code or technical jargon, include visualizations
#  - Focus on metrics that are relevant for the business
#  - Talk about feature importance and how to interpret results
#  - Explain the strengths and limitations of the model
#  - Explain how the model can be improved over time


#  4. Kaggle Submissions
# The final output is a CSV file with predictions. You format it exactly as the competition requires and submit it to get your final score.



## Model Deployment
# Once the model is trained and evaluated, it needs to be handed off to a software developer or ML engineer.
#  -> Putting it into Production: The model is integrated into an existing software system (like a web app or a mobile app)
#     so real users can interact with it.
#  -> Monitoring: The model is not "set and forget." You have to monitor its performance over time. If the real-world data changes
#     (e.g., customer behavior shifts), the model will become less accurate.
#  -> Iteration: You must retrain and update the model from time to time to keep it performing well.

## This is the final box on the diagram. You have officially completed the entire Machine Learning Lifecycle! 🎉🚀
```
